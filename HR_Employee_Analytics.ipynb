```python
from datetime import datetime
current_date = pd.Timestamp.today()
df['Calculated_Years'] = (
    (current_date - df['Join_Date']).dt.days / 365.25
)
df[['Join_Date', 'Years_at_Company', 'Calculated_Years']].head(10)
```

```python
df.drop(columns=['Calculated_Years'], inplace=True)
df.columns
```

```python
missing_salary = df[df['Salary_INR'].isnull()]
missing_performance = df[df['Performance_Rating'].isnull()]
print("Rows with missing Salary:")
display(missing_salary)
print("\nRows with missing Performance Rating:")
display(missing_performance)
```

```python
print("Salary median:", df['Salary_INR'].median())
print("Salary mean:", df['Salary_INR'].mean())
print("\nPerformance Rating:")
print(df['Performance_Rating'].value_counts())
```

```python
df['Salary_INR'].fillna(df['Salary_INR'].median(), inplace=True)
df['Performance_Rating'].fillna(
    df['Performance_Rating'].mode()[0],
    inplace=True
)
df.isnull().sum()
```

```python
df['Salary_INR'] = df['Salary_INR'].fillna(df['Salary_INR'].median())
df['Performance_Rating'] = df['Performance_Rating'].fillna(
    df['Performance_Rating'].mode()[0]
)
df.isnull().sum()
```

```python
department_status = df.groupby(['Department', 'Employment_Status']).size().unstack(fill_value=0)
department_status['Total'] = department_status.sum(axis=1)
department_status['Attrition_Rate'] = (
    (department_status['Resigned'] + department_status['Terminated'])
    / department_status['Total'] * 100
)
department_status.sort_values('Attrition_Rate', ascending=False)
```

```python
performance_status = df.groupby(
    ['Performance_Rating', 'Employment_Status']
).size().unstack(fill_value=0)
performance_status['Total'] = performance_status.sum(axis=1)
performance_status['Attrition_Rate'] = (
    (performance_status['Resigned'] + performance_status['Terminated'])
    / performance_status['Total'] * 100
)
performance_status.sort_values('Attrition_Rate', ascending=False)
```

```python
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[20, 30, 40, 50, 60],
    labels=['21-30', '31-40', '41-50', '51-60']
)
age_status = df.groupby(
    ['Age_Group', 'Employment_Status'],
    observed=False
).size().unstack(fill_value=0)
age_status['Total'] = age_status.sum(axis=1)
age_status['Attrition_Rate'] = (
    (age_status['Resigned'] + age_status['Terminated'])
    / age_status['Total'] * 100
)
age_status
```

```python
age_salary = df.groupby('Age_Group', observed=False)['Salary_INR'].agg(
    ['count', 'mean', 'median']
)
age_salary
```
